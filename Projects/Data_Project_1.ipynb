```python
"""
In order to use properly under runtime hit run all, then provide
a number (1 or 2) depending on whether the data will be accesed using an API
or it is stored locally.

API:
if running on an API provide a valid url

Local:
File must be uploaded onto colab
then just type in the file name (ex: all_day.csv)

After fetching is complete provide the format
you would like the data to be in.

"""

# RETRIEVE

import requests as r
import pandas as pd
from io import StringIO
import sqlite3

# ask if retrieving file from API or Locally
print("Enter Choice: \n 1) API \n 2) Local")

while True:

  source_choice = input("Enter input here: ")

  if source_choice != '1' and source_choice != '2':
    print("Please enter a valid choice (1 or 2)")
  else:
    print("Choice sucessfully documented, please enter your URL or file path")
    break

path = input()
print("\n")

if source_choice == '1': # API

  try:
    response = r.get(path)
    response.raise_for_status()
    format = response.headers['Content-Type']

    if "/csv" in format:
      csv_r = StringIO(response.text)
      df = pd.read_csv(csv_r)
      print("Here is your data in data frame format:")
      print("-" * 39)
      print(df.head())

    elif "/json" in format:
      df = pd.DataFrame(response.json())
      print("Here is your data in data frame format:")
      print("-" * 39)
      print(df.head())

    else:
      print("API doesn't retrieve csv or json file, please use a valid format")

  except r.exceptions.RequestException as exception:
    print("Failed to retrieve file, error code: " + str(exception))

else: # option 2, Local

  try:
    if path.lower().endswith("csv"):
      df = pd.read_csv(path)
      print("Here is your data in data frame format:")
      print("-" * 39)
      print(df.head())

    elif path.lower().endswith("json"):
      df = pd.read_json(path)
      print("Here is your data in data frame format:")
      print("-" * 39)
      print(df.head())

    else:
      print("Please enter a path with a valid file type.")

  except Exception as e:
    print("Failed to retrieve file from local path, error code: " + str(e))
```

```python
# MODIFY

print(f"Records and columns pre-modification: [{len(df)},{len(df.columns)}]")

# if statements to make sure nothing is removed that doesn't exist
if len(df) > 1 and len(df.columns) > 1:
  df = df.iloc[:-1, :]
  df = df.iloc[:, :-1]
else:
  df.loc[len(df)] = ["NA"] * len(df.columns)
  df.index = df.index + 1
  df = df.sort_index()

print(f"Records and columns post-modification: [{len(df)},{len(df.columns)}]")
```

```python
# CONVERT/STORE

print("What file type would you like to convert to?: \n 1) csv \n 2) json \n 3) sql")

while True:

  source_choice = input("Enter input here: ")

  if source_choice != '1' and source_choice != '2' and source_choice != '3':
    print("Please enter a valid choice (1, 2, 3)")
  else:
    print("Choice sucessfully documented, data has been saved to your local machine.")
    break

match source_choice:
  case '1':
    df.to_csv("data.csv")
  case '2':
    df.to_json("data.json")
  case '3':
    connection = sqlite3.connect("sql_database.sql")
    df.to_sql("data", connection, if_exists = "replace", index = False)
    connection.close()
```
